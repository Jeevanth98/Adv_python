```python

#TASK - 01
#Thread creation using threading module - No loop

import threading
import time

start=time.perf_counter()
def do_something():
    print(f'sleeping for 5 seconds')
    time.sleep(5)
    print('Done Sleeping')

t1=threading.Thread(target=do_something)   #Thread 1
t1.start()     #Starts the thread execution
t1.join()

t2=threading.Thread(target=do_something)   #Thread 2
t2.start()
t2.join()      #Tells to wait until the current thread is completed

finish=time.perf_counter()
print(f'Finished in {round(finish-start,2)} second(s)')

```

```python
#TASK-03
#Passing arguments to the function during thread creation

import threading
import time

start=time.perf_counter()

def do_something(seconds):     # Now the function accepts argument "seconds"
    print(f'sleeping for {seconds} seconds')
    time.sleep(seconds)        # sleeps for {seconds} second
    print('Done Sleeping')

threads=[]                     #to store the threads
for _ in range(10):
    t1=threading.Thread(target=do_something, args=[4])    #passing arguments using args keyword
                                                           #which accepts a list of arguments

    t1.start()
    threads.append(t1)

for i in threads:
    i.join()

finish=time.perf_counter()
print(f'Finished in {round(finish-start,2)} second(s)')
```

```python
#TASK-04
#Thread creation using concurrent futures module

import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds):
    print(f'sleeping for {seconds} seconds')
    time.sleep(seconds)
    return 'Done Sleeping'

#with - context manager
#ThreadPoolExecutor() is the class to be instantiated to create threads
with concurrent.futures.ThreadPoolExecutor() as executor:
    f1=executor.submit(do_something,1.5)  #f1 captures the return statement of the do_something func (future object)
    f2=executor.submit(do_something,1.5)
    print(f1.result())       #result function to print the stored returned result
    print(f2.result())

finish=time.perf_counter()
print(f'Finished on {round(finish-start,2)} second(s)')
```

```python
#TASK - 05
#Thread creation using list comprehension

import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds):
    print(f'sleeping for {seconds} seconds')
    time.sleep(seconds)
    return 'Done Sleeping'

with concurrent.futures.ThreadPoolExecutor() as executor:
    res=[executor.submit(do_something,1.5) for _ in range(5)] #list comprehension

    for f in concurrent.futures.as_completed(res):
        print(f.result())

finish=time.perf_counter()
print(f'Finished on {round(finish-start,2)} second(s)')
```

```python
#TASK - 06
#passing list of arguments to the function

import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds):
    print(f'sleeping for {seconds} seconds')
    time.sleep(seconds)
    return 'Done Sleeping'

with concurrent.futures.ThreadPoolExecutor() as executor:
    secs=[2,3,4,5,6]    #list of seconds to be passed as args to the function
    res=[executor.submit(do_something,sec) for sec in secs]    #list comprehension

    for f in concurrent.futures.as_completed(res):
        print(f.result())

finish=time.perf_counter()
print(f'Finished on {round(finish-start,2)} second(s)')
```

```python
#TASK - 07
#Using map() method instead of submit

import concurrent.futures
import time

start=time.perf_counter()

def do_something(seconds):
    print(f'sleeping for {seconds} seconds')
    time.sleep(seconds)
    return 'Done Sleeping'

with concurrent.futures.ThreadPoolExecutor() as executor:
    secs=[2,3,4,5,6]    #list of seconds to be passed as args to the function
    res=executor.map(do_something,secs)
#map doesnt hold a future object, it holds the result in the order they were started

    for result in res:
        print(result)

finish=time.perf_counter()
print(f'Finished on {round(finish-start,2)} second(s)')
```

```python
#TASK - 08
#Downloading high-resolution files with and
#without using threads and
#comparing the time taken

import os
import time
import urllib.request
from concurrent.futures import ThreadPoolExecutor

audio_urls = [
    "https://w3schools.com",
    "https://filesamples.com",
    "https://soundhelix.com",
    "https://dogphilosophy.net",
    "https://archive.org",
    "https://wikimedia.org",
    "https://utexas.edu",
    "https://uu.nl",
    "https://nasa.gov",
    "https://mozilla.net"
]

def download_file(url, prefix, index):
    # Deduce extension/format safely
    ext = ".ogg" if ".ogg" in url.lower() else ".mp3"
    filename = f"{prefix}_audio{index}{ext}"

    try:
        req = urllib.request.Request(
            url,
            headers={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"}
        )

        with urllib.request.urlopen(req, timeout=10) as response, open(filename, 'wb') as out_file:
            out_file.write(response.read())

        size_bytes = os.path.getsize(filename)
        size_kb = size_bytes / 1024
        return filename, size_kb, True

    except Exception as e:
        # Graceful handling for intermittent network timeouts or firewall filters
        return f"Error downloading index {index}: {e}", 0.0, False


def run_sequential(urls):
    print("\n" + "=" * 60)
    print("🚀 METHOD 1: RUNNING SEQUENTIAL DOWNLOAD (No Threads)")
    print("=" * 60)
    start = time.perf_counter()
    results = []
    
    for i, url in enumerate(urls, start=1):
        domain = url.split('/')[2]
        print(f"[{i}/{len(urls)}] Target Host -> {domain}")
        name, size, success = download_file(url, "seq", i)
        if success:
            results.append((name, size))

    end = time.perf_counter()
    duration = end - start

    print("\n📥 Sequential Storage Receipt:")
    for name, size in results:
        print(f"  • {name} | Size: {size:.2f} KB")
    print(f"⏱ Total Sequential Execution Duration: {duration:.4f} seconds")
    return duration


def run_threaded(urls):
    print("\n" + "=" * 60)
    print("🚀 METHOD 2: RUNNING PARALLEL DOWNLOAD (ThreadPoolExecutor)")
    print("=" * 60)
    start = time.perf_counter()

    # max_workers=10 enables all 10 downloads to run at the exact same millisecond
    with ThreadPoolExecutor(max_workers=10) as executor:
        futures = [
            executor.submit(download_file, url, "thread", i)
            for i, url in enumerate(urls, start=1)
        ]
        results = [f.result() for f in futures]

    end = time.perf_counter()
    duration = end - start

    print("\n📥 Threaded Storage Receipt:")
    for name, size, success in results:
        if success:
            print(f"  • {name} | Size: {size:.2f} KB")
    print(f"⏱ Total Threaded Execution Duration: {duration:.4f} seconds")
    return duration


if __name__ == "__main__":
    print("Initializing upgraded 10-domain cross-performance metric tracking...")

    seq_time = run_sequential(audio_urls)
    thread_time = run_threaded(audio_urls)

    # Calculate performance difference metrics
    time_saved = seq_time - thread_time
    efficiency_gain = (time_saved / seq_time) * 100 if seq_time > 0 else 0

    print("\n" + "=" * 17 + " FINAL METRIC ANALYSIS COMPARISON " + "=" * 17 + "=")
    print("=" * 68)
    print(f"📌 Baseline Sequential Method Processing Time: {seq_time:.4f} seconds")
    print(f"⚡ Optimized Parallel Threading Processing Time: {thread_time:.4f} seconds")
    print(f"📉 Total Network Wait Reduction: {time_saved:.4f} seconds faster")
    print(f"📊 Script Performance Optimization Rating : {efficiency_gain:.1f}% improvement")
    print("=" * 68)

```

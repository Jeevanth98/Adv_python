```python
class Student:
    def __init__(self,name,id):
        self.name=name
        self.id=id
        self.scores=[]

    def add_score(self,score):
        self.scores.append(score)

    def calc_avg(self):
        if len(self.scores)==0:
            return 0
        return sum(self.scores)/len(self.scores)

    def final_grade(self):
        if self.calc_avg()>=60:
            return "Pass"
        return "Fail"
```

```python
#1.Object Creation
student1 = Student("Alice", 101)
print(student1.name)        
print(student1.id)   
print(student1.scores)       
```

```python
#2.Adding Scores
student1.add_score(75)
student1.add_score(85)
student1.add_score(55)
print(student1.scores)
```

```python
#3.Calculate Average
print(student1.calc_avg())
```

```python
print(round(student1.calc_avg(),2))
```

```python
#4.Final Grade (Pass Case)
print(student1.final_grade())
```

```python
#5.Final Grade(Fail Case)
student2=Student("Bob", 102)
student2.add_score(40)
student2.add_score(50)
print(student2.calc_avg()) 
print(student2.final_grade()) 
```

```python
#6.Empty Scores Case
student3 = Student("Charlie", 103)
print(student3.calc_avg())
print(student3.final_grade()) 
```

```python
class Gene:
    def __init__(self, gene_id, seq):
        print("I'm a new Gene object!")
        print(f"My constructor got a param: {gene_id}")
        print("Assign that param to my id instance variable...")
        self.id = gene_id
        print("Similarly, assigning to my sequence instance variable...")
        assert "U" not in seq.upper(), "Sorry, no RNA allowed."
        self.seq = seq.upper()

    def print_id(self):
        print("My id is:", self.id)

    def print_seq(self):
        print("My sequence is:", self.seq)

    def base_composition(self, base):
        return self.seq.count(base.upper())

    def gc_content(self):
        gc_count = self.seq.count("G") + self.seq.count("C")
        return gc_count / len(self.seq)

    def get_seq(self):
        return self.seq

    def set_seq(self, new_seq):
        assert "U" not in new_seq.upper(), "Sorry, no RNA allowed."
        self.seq = new_seq.upper()
```

```python
#1.Object Creation
gene1 = Gene(101, "ATGCGT")
```

```python
#2.Print ID and Sequence
gene1.print_id()
gene1.print_seq()
```

```python
#3.Base Composition
print("Base Composition of A: ",gene1.base_composition("A"))
print("Base Composition of G: ",gene1.base_composition("G"))
print("Base Composition of T: ",gene1.base_composition("T"))
```

```python
#4.GC Content
print(gene1.gc_content())
```

```python
#5.Get Sequence
print(gene1.get_seq())
```

```python
#6.Set Sequence(Valid DNA)
gene1.set_seq("ATATGC")
print(gene1.get_seq())
```

```python
#7.Set Sequence(Invalid RNA)
gene1.set_seq("AUGC")
```
